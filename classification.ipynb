```python
import numpy as np
import pandas as pd

# Select numerical columns
num_cols = df_model.select_dtypes(include=np.number)

# Exclude binary, target, and categorical label columns
exclude_cols = [
    'Is_Holiday',
    'Accident_Occurred',
    'Vehicles_Involved',
    'Accident_Severity'
]

# Keep only continuous numerical columns
continuous_cols = [
    col for col in num_cols.columns
    if col not in exclude_cols
    and num_cols[col].nunique() > 2
]

# Detect outliers using IQR
outlier_counts = {}

for col in continuous_cols:
    q1 = df_model[col].quantile(0.25)
    q3 = df_model[col].quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    count = (
        (df_model[col] < lower_bound) |
        (df_model[col] > upper_bound)
    ).sum()

    outlier_counts[col] = count

# Display results
outlier_summary = pd.DataFrame(
    list(outlier_counts.items()),
    columns=['Column', 'Outlier_Count']
).sort_values('Outlier_Count', ascending=False)

print(outlier_summary.to_string(index=False))
```

```python
import numpy as np
import pandas as pd

time_cols = ['Baseline_Expected_Response_Time_min', 'Actual_Response_Time_min', 'Road_Closure_Duration_min']

for col in time_cols:
    
    df_model[f'{col}_log'] = np.log1p(df_model[col])

log_summary = []

for col in time_cols:
    log_col_name = f'{col}_log'
    
    skew_val = df_model[log_col_name].skew()
    
    
    q1 = df_model[log_col_name].quantile(0.25)
    q3 = df_model[log_col_name].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    
    outliers_count = ((df_model[log_col_name] < lower_bound) | (df_model[log_col_name] > upper_bound)).sum()
    
    log_summary.append({
        'Column': log_col_name,
        'New_Skewness': skew_val,
        'Remaining_Outliers_After_Log': outliers_count
    })

df_log_summary = pd.DataFrame(log_summary)
print(df_log_summary.to_string(index=False))
```

```python
def cap_outliers_iqr(df_model, column, lower_limit=0.0):
    q1 = df_model[column].quantile(0.25)
    q3 = df_model[column].quantile(0.75)
    iqr = q3 - q1
    
    lower_bound = max(q1 - 1.5 * iqr, lower_limit)
    upper_bound = q3 + 1.5 * iqr
    
    df_model[column] = df_model[column].clip(lower=lower_bound, upper=upper_bound)
    return lower_bound, upper_bound

lower_Recorded_Speed_kmh, upper_Recorded_Speed_kmh = cap_outliers_iqr(df_model, 'Recorded_Speed_kmh', lower_limit=0.0)
print(f"Recorded_Speed_kmh capped between: {lower_Recorded_Speed_kmh:.3f} and {upper_Recorded_Speed_kmh:.2f}")

lower_Vehicle_Age_yrs, upper_Vehicle_Age_yrs = cap_outliers_iqr(df_model, 'Vehicle_Age_yrs', lower_limit=0.0)
print(f"upper_Vehicle_Age_yrs capped between: {lower_Vehicle_Age_yrs:.2f} and {upper_Vehicle_Age_yrs:.2f}")

lower_Driver_Experience_yrs, upper_Driver_Experience_yrs = cap_outliers_iqr(df_model, 'Driver_Experience_yrs', lower_limit=0.0)
print(f"Driver_Experience_yrs capped between: {lower_Driver_Experience_yrs:.2f} and {upper_Driver_Experience_yrs:.2f}")

lower_Driver_Age, upper_Driver_Age = cap_outliers_iqr(df_model, 'Driver_Age', lower_limit=16.0)
print(f"Driver_Age capped between: {lower_Driver_Age:.2f} and {upper_Driver_Age:.2f}")
```

```python

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

models = {
    "Logistic Regression": Pipeline([
        ('imputer', SimpleImputer(strategy='median')),
        ('scaler', StandardScaler()),
        ('model', LogisticRegression(
            max_iter=2000,
            class_weight='balanced'
        ))
    ]),

    "Decision Tree": Pipeline([
        ('imputer', SimpleImputer(strategy='median')),
        ('model', DecisionTreeClassifier(
            max_depth=6,
            min_samples_split=50,
            min_samples_leaf=20,
            class_weight='balanced',
            random_state=42
        ))
    ]),

    "Random Forest": Pipeline([
        ('imputer', SimpleImputer(strategy='median')),
        ('model', RandomForestClassifier(
            n_estimators=100,
            max_depth=10,
            min_samples_split=50,
            min_samples_leaf=20,
            class_weight='balanced',
            random_state=42,
            n_jobs=-1
        ))
    ])
}
```
